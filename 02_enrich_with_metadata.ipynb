```python
# Since the repository is public, the storage account name and key has been removed from all the notebooks 

storage_account_name = ""
storage_account_key = ""
spark.conf.set(
f"fs.azure.account.key.{storage_account_name}.dfs.core.windows.net",
storage_account_key
)
```

```python
clean_reviews_path = "abfss://processed@.dfs.core.windows.net/clean_reviews/"
clean_reviews_df = spark.read.parquet(clean_reviews_path)
```

```python
metadata_path = "abfss://raw@.dfs.core.windows.net/meta_Electronics_fixed.json"
metadata_df = spark.read.json(metadata_path)
```

```python
metadata_df = metadata_df.select(
"asin",
"title",
"brand",
"price"
)
```

```python
enriched_df = clean_reviews_df.join(
metadata_df,
on="asin",
how="left"
)
```

```python
enriched_reviews_path = "abfss://processed@.dfs.core.windows.net/enriched_reviews/"
enriched_df.write.mode("overwrite").parquet(enriched_reviews_path)
```
